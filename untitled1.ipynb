## 1. Environment Setup and Data Loading

```python
import torch
import torch.nn as nn
import math
import pandas as pd
import kagglehub
from transformers import AutoTokenizer

path = kagglehub.dataset_download("gowrishankarp/newspaper-text-summarization-cnn-dailymail")

train_df = pd.read_csv(f"{path}/cnn_dailymail/train.csv").head(2000)
val_df = pd.read_csv(f"{path}/cnn_dailymail/validation.csv").head(200)
test_df = pd.read_csv(f"{path}/cnn_dailymail/test.csv").head(200)
```

## 2. Tokenizer and Dataset Initialization

```python
tokenizer = AutoTokenizer.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token

class SummarizationDataset(torch.utils.data.Dataset):
    def __init__(self, tokenizer, articles, summaries, max_len=256):
        self.tokenizer = tokenizer
        self.articles = articles
        self.summaries = summaries
        self.max_len = max_len

    def __len__(self):
        return len(self.articles)

    def __getitem__(self, idx):
        encoder_input = self.tokenizer(
            str(self.articles[idx]), 
            max_length=self.max_len, 
            padding="max_length", 
            truncation=True, 
            return_tensors="pt"
        )
        decoder_input = self.tokenizer(
            str(self.summaries[idx]), 
            max_length=self.max_len, 
            padding="max_length", 
            truncation=True, 
            return_tensors="pt"
        )
        return {
            "encoder_input_ids": encoder_input["input_ids"].squeeze(0),
            "decoder_input_ids": decoder_input["input_ids"].squeeze(0),
            "labels": decoder_input["input_ids"].squeeze(0).clone(),
        }

MAX_LEN = 256
BATCH_SIZE = 8

train_dataset = SummarizationDataset(tokenizer, train_df["article"].tolist(), train_df["highlights"].tolist(), MAX_LEN)
val_dataset = SummarizationDataset(tokenizer, val_df["article"].tolist(), val_df["highlights"].tolist(), MAX_LEN)

train_loader = torch.utils.data.DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader = torch.utils.data.DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)
```

## 3. Transformer Architecture

```python
class MultiHeadAttention(nn.Module):
    def __init__(self, d_model, num_heads):
        super().__init__()
        self.num_heads = num_heads
        self.head_dim = d_model // num_heads
        self.W_Q = nn.Linear(d_model, d_model)
        self.W_K = nn.Linear(d_model, d_model)
        self.W_V = nn.Linear(d_model, d_model)
        self.W_O = nn.Linear(d_model, d_model)

    def forward(self, query, key, value, mask=None):
        batch_size = query.shape[0]
        Q = self.W_Q(query).view(batch_size, -1, self.num_heads, self.head_dim).transpose(1, 2)
        K = self.W_K(key).view(batch_size, -1, self.num_heads, self.head_dim).transpose(1, 2)
        V = self.W_V(value).view(batch_size, -1, self.num_heads, self.head_dim).transpose(1, 2)

        scores = torch.matmul(Q, K.transpose(-2, -1)) / math.sqrt(self.head_dim)
        if mask is not None:
            scores = scores.masked_fill(mask == 0, float('-inf'))

        attention_weights = torch.softmax(scores, dim=-1)
        attention_output = torch.matmul(attention_weights, V).transpose(1, 2).contiguous().view(batch_size, -1, self.num_heads * self.head_dim)
        return self.W_O(attention_output), attention_weights

class FeedForwardNetwork(nn.Module):
    def __init__(self, d_model, d_ff):
        super().__init__()
        self.linear_1 = nn.Linear(d_model, d_ff)
        self.relu = nn.ReLU()
        self.linear_2 = nn.Linear(d_ff, d_model)

    def forward(self, x):
        return self.linear_2(self.relu(self.linear_1(x)))

class EncoderBlock(nn.Module):
    def __init__(self, d_model, num_heads, d_ff, dropout_rate=0.1):
        super().__init__()
        self.self_attention = MultiHeadAttention(d_model, num_heads)
        self.norm_1 = nn.LayerNorm(d_model)
        self.dropout_1 = nn.Dropout(dropout_rate)
        self.ffn = FeedForwardNetwork(d_model, d_ff)
        self.norm_2 = nn.LayerNorm(d_model)
        self.dropout_2 = nn.Dropout(dropout_rate)

    def forward(self, x, mask=None):
        attention_output, _ = self.self_attention(x, x, x, mask)
        x = self.norm_1(x + self.dropout_1(attention_output))
        ffn_output = self.ffn(x)
        x = self.norm_2(x + self.dropout_2(ffn_output))
        return x

class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        self.register_buffer("pe", pe.unsqueeze(0))

    def forward(self, x):
        return x + self.pe[:, :x.size(1)]

class Encoder(nn.Module):
    def __init__(self, vocab_size, d_model, num_heads, d_ff, num_layers, dropout_rate=0.1, max_len=5000):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, d_model)
        self.positional_encoding = PositionalEncoding(d_model, max_len)
        self.dropout = nn.Dropout(dropout_rate)
        self.layers = nn.ModuleList([EncoderBlock(d_model, num_heads, d_ff, dropout_rate) for _ in range(num_layers)])
        self.norm = nn.LayerNorm(d_model)

    def forward(self, src, src_mask=None):
        x = self.dropout(self.positional_encoding(self.embedding(src)))
        for layer in self.layers:
            x = layer(x, src_mask)
        return self.norm(x)

class DecoderBlock(nn.Module):
    def __init__(self, d_model, num_heads, d_ff, dropout_rate=0.1):
        super().__init__()
        self.self_attention = MultiHeadAttention(d_model, num_heads)
        self.norm_1 = nn.LayerNorm(d_model)
        self.dropout_1 = nn.Dropout(dropout_rate)
        self.cross_attention = MultiHeadAttention(d_model, num_heads)
        self.norm_2 = nn.LayerNorm(d_model)
        self.dropout_2 = nn.Dropout(dropout_rate)
        self.ffn = FeedForwardNetwork(d_model, d_ff)
        self.norm_3 = nn.LayerNorm(d_model)
        self.dropout_3 = nn.Dropout(dropout_rate)

    def forward(self, tgt, encoder_output, tgt_mask=None, src_mask=None):
        attention_output, _ = self.self_attention(tgt, tgt, tgt, tgt_mask)
        tgt = self.norm_1(tgt + self.dropout_1(attention_output))
        cross_attention_output, _ = self.cross_attention(tgt, encoder_output, encoder_output, src_mask)
        tgt = self.norm_2(tgt + self.dropout_2(cross_attention_output))
        ffn_output = self.ffn(tgt)
        tgt = self.norm_3(tgt + self.dropout_3(ffn_output))
        return tgt

class Decoder(nn.Module):
    def __init__(self, vocab_size, d_model, num_heads, d_ff, num_layers, dropout_rate=0.1, max_len=5000):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, d_model)
        self.positional_encoding = PositionalEncoding(d_model, max_len)
        self.dropout = nn.Dropout(dropout_rate)
        self.layers = nn.ModuleList([DecoderBlock(d_model, num_heads, d_ff, dropout_rate) for _ in range(num_layers)])
        self.norm = nn.LayerNorm(d_model)
        self.output_layer = nn.Linear(d_model, vocab_size)

    def forward(self, tgt, encoder_output, tgt_mask=None, src_mask=None):
        x = self.dropout(self.positional_encoding(self.embedding(tgt)))
        for layer in self.layers:
            x = layer(x, encoder_output, tgt_mask, src_mask)
        return self.output_layer(self.norm(x))

class Transformer(nn.Module):
    def __init__(self, vocab_size, d_model, num_heads, d_ff, num_layers, dropout_rate=0.1, max_len=5000):
        super().__init__()
        self.encoder = Encoder(vocab_size, d_model, num_heads, d_ff, num_layers, dropout_rate, max_len)
        self.decoder = Decoder(vocab_size, d_model, num_heads, d_ff, num_layers, dropout_rate, max_len)

    def forward(self, src, tgt, src_mask=None, tgt_mask=None):
        encoder_output = self.encoder(src, src_mask)
        return self.decoder(tgt, encoder_output, tgt_mask, src_mask)
```

## 4. Model Initialization

```python
d_model = 128 
num_heads = 4
d_ff = 256
num_layers = 2
dropout_rate = 0.1
vocab_size = tokenizer.vocab_size

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = Transformer(vocab_size, d_model, num_heads, d_ff, num_layers, dropout_rate).to(device)

criterion = nn.CrossEntropyLoss(ignore_index=tokenizer.pad_token_id)
optimizer = torch.optim.Adam(model.parameters(), lr=0.0001, betas=(0.9, 0.98), eps=1e-9)
```

## 5. Training Loop

```python
def create_look_ahead_mask(size, device):
    return torch.tril(torch.ones(size, size, dtype=torch.bool, device=device))

def train_model(model, train_loader, val_loader, optimizer, criterion, device, num_epochs=1):
    for epoch in range(num_epochs):
        model.train()
        total_loss = 0
        for batch_idx, batch in enumerate(train_loader):
            encoder_input_ids = batch["encoder_input_ids"].to(device)
            decoder_input_ids = batch["decoder_input_ids"].to(device)
            labels = batch["labels"].to(device)

            encoder_mask = (encoder_input_ids != tokenizer.pad_token_id).unsqueeze(1).unsqueeze(2).to(device)
            
            tgt_seq_len = decoder_input_ids.size(1) - 1
            decoder_padding_mask = (decoder_input_ids[:, :-1] != tokenizer.pad_token_id).unsqueeze(1).unsqueeze(2).to(device)
            decoder_causal_mask = create_look_ahead_mask(tgt_seq_len, device).unsqueeze(0).unsqueeze(1)
            combined_decoder_mask = decoder_padding_mask & decoder_causal_mask

            optimizer.zero_grad()

            output = model(
                src=encoder_input_ids,
                tgt=decoder_input_ids[:, :-1],
                src_mask=encoder_mask,
                tgt_mask=combined_decoder_mask
            )

            output_flat = output.contiguous().view(-1, output.size(-1))
            labels_flat = labels[:, 1:].contiguous().view(-1)

            loss = criterion(output_flat, labels_flat)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()

            if batch_idx % 50 == 0:
                print(f"Epoch {epoch+1} | Batch {batch_idx}/{len(train_loader)} | Loss: {loss.item():.4f}")

        print(f"Epoch {epoch+1} Completed | Average Train Loss: {total_loss / len(train_loader):.4f}")

train_model(model, train_loader, val_loader, optimizer, criterion, device, num_epochs=1)
```

```python
def generate_summary(model, tokenizer, text, max_len=256, max_gen_len=60, temperature=0.7, top_k=40, repetition_penalty=2.0, device="cpu"):
    model.eval()
    
    inputs = tokenizer(
        str(text),
        max_length=max_len,
        padding="max_length",
        truncation=True,
        return_tensors="pt"
    )
    
    src = inputs["input_ids"].to(device)
    src_mask = (src != tokenizer.pad_token_id).unsqueeze(1).unsqueeze(2).to(device)

    tgt_tokens = [tokenizer.eos_token_id]

    with torch.no_grad():
        encoder_out = model.encoder(src, src_mask)
        
        for _ in range(max_gen_len):
            tgt_tensor = torch.tensor([tgt_tokens], dtype=torch.long, device=device)
            seq_len = tgt_tensor.size(1)
            
            causal_mask = torch.tril(torch.ones(seq_len, seq_len, dtype=torch.bool, device=device)).unsqueeze(0).unsqueeze(1)
            
            output = model.decoder(tgt_tensor, encoder_out, tgt_mask=causal_mask, src_mask=src_mask)
            logits = output[:, -1, :] / temperature
            
            for token_id in set(tgt_tokens):
                if logits[0, token_id] < 0:
                    logits[0, token_id] *= repetition_penalty
                else:
                    logits[0, token_id] /= repetition_penalty

            top_values, top_indices = torch.topk(logits, top_k)
            probs = torch.softmax(top_values, dim=-1)
            next_token = top_indices[0, torch.multinomial(probs, num_samples=1).item()].item()
            
            tgt_tokens.append(next_token)
            
            if next_token == tokenizer.eos_token_id:
                break

    return tokenizer.decode(tgt_tokens[1:], skip_special_tokens=True).strip()

sample_idx = 0
article = test_df["article"].iloc[sample_idx]
target = test_df["highlights"].iloc[sample_idx]

summary = generate_summary(
    model=model,
    tokenizer=tokenizer,
    text=article,
    max_len=MAX_LEN,
    max_gen_len=60,
    temperature=0.8,
    top_k=50,
    repetition_penalty=2.5,
    device=device
)

print("ARTICLE:")
print(article[:350], "...\n")
print("TARGET:")
print(target, "\n")
print("GENERATED:")
print(summary)
```
